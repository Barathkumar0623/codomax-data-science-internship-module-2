## Data Analysis with Python

### Objectives
- Import CSV datasets using Pandas

- Understand rows and columns

```python
import pandas as pd

from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_csv("CovidDeaths.csv")

df.head()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print("Column Names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:", df.duplicated().sum())
```

```python
print("Number of Unique Locations:", df["location"].nunique())
```

```python
print("Sample Locations:")
print(df["location"].unique()[:20])
```

```python
print("Total Cases:", df["total_cases"].max())
print("Total Deaths:", df["total_deaths"].max())
```
