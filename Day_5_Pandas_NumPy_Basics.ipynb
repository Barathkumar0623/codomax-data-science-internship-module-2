```python
import numpy as np
numbers = np.array([10,20,30,40,50])

print("Arrays:" ,numbers)
print("Shape:", numbers.shape)
print("Size:" ,numbers.size)
print("Mean:", numbers.mean())
print("Maximum:", numbers.max())
print("Minimum:", numbers.min())
print("Sum:", numbers.sum())
```

Pandas

```python
import pandas as pd

data = {
  "Name" : ["Arun","Bala","Charan","Deepak","Elango"],
  "Age"  :  [21,22,23,24,25]
}

df = pd.DataFrame(data)
print(df)
```

```python
print("First 5 rows:")
display(df.head())

print("Data Information:")
df.info()

print("Statistical Summary:")
display(df.describe())
```

```python
data = {
    "Name": ["Arun", "Bala", "Charan", "Dinesh", "Eshan"],
    "Age": [21, 22, 20, 23, 21],
    "Score": [85, 78, 92, 67, 88]
}

df = pd.DataFrame(data)

df

print("Average Score:", df["Score"].mean())
print("Highest Score:", df["Score"].max())
print("Lowest Score:", df["Score"].min())
```

```python
print("Data Types:")
print(df.dtypes)
```
